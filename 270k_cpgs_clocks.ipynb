```python
import pandas as pd 
import pyaging as pya
```

```python
path_clock_data = "E:/YandexDisk/pydnameth/datasets/pyaging"
path_270k = "E:/YandexDisk/pydnameth/datasets/GPL34394"
path_clocks_names = "E:/YandexDisk/bbd/attacks_on_clocks"

clock_names_file = open(f"{path_clocks_names}/clocks_list.txt", "r") 
clocks_names = clock_names_file.read().split("\n") 
clock_names_file.close() 

path_clock_meta = "E:/YandexDisk/bbd/immunology/003_EpImAge/epi"
clock_meta = pd.read_excel(f"{path_clock_meta}/clocks_meta.xlsx")
clock_meta = clock_meta.set_index('Model ID')

anno_270k = pd.read_excel(f"{path_270k}/msa.xlsx")
cpgs_270k = list(anno_270k['Name'].values)
```

```python
df = pd.DataFrame(index=['EpImAge'].extend(clocks_names))

cpgs_EpImAge = pd.read_excel("E:/YandexDisk/bbd/immunology/003_EpImAge/imp_source(imm)_method(knn)_params(5)/no_harm/mrmr_100/EpImAge/cpgs_selected.xlsx")
cpgs_EpImAge = list(cpgs_EpImAge['CpG'].values)
df.loc['EpImAge', 'Total number of CpGs'] = int(len(cpgs_EpImAge))
differ_cpgs = list(set(cpgs_EpImAge).difference(set(cpgs_270k)))
df.loc['EpImAge', 'Number of CpGs missed in 270k'] = int(len(differ_cpgs))
df.loc['EpImAge', 'Persentage of missed CpGs'] = len(differ_cpgs) / len(cpgs_EpImAge)
df.loc['EpImAge', 'Year'] = ''

logger = pya.logger.Logger('test_logger')
device = 'cpu'
indent_level = 1

for clock_name in clocks_names:
    clock = pya.pred.load_clock(clock_name, device, path_clock_data, logger, indent_level=indent_level)
    clock_features = clock.features
    cpgs_features = []
    for feat in clock_features:
        if feat.startswith("cg") or feat.startswith("ch") or feat.startswith("rs"):
            cpgs_features.append(feat)
    df.loc[clock_name, 'Total number of CpGs'] = int(len(cpgs_features))
    differ_cpgs = list(set(cpgs_features).difference(set(cpgs_270k)))
    df.loc[clock_name, 'Number of CpGs missed in 270k'] = int(len(differ_cpgs))
    df.loc[clock_name, 'Persentage of missed CpGs'] = len(differ_cpgs) / len(cpgs_features)
    df.loc[clock_name, 'Year'] = int(clock_meta.loc[clock_name, 'Year'])

df.to_excel("Num_CpGs.xlsx", index_label='Clock')
```
