```python
import pandas as pd
import pyaging as pya
```

```python
logger = pya.logger.Logger('test_logger')
device = 'cpu'
dir = 'pyaging_data'
indent_level = 1

clock = pya.pred.load_clock('GrimAge', device, dir, logger, indent_level=indent_level)
feats = list(clock.features)
```

```python
path = "C:/Git/bioTest/test_dnam_data/"
pheno = pd.read_excel(f"{path}/pheno_funnorm.xlsx", index_col=0)
pheno.index = pheno.index.astype(str)

betas = pd.read_pickle(f"{path}/betas_funnorm.pkl")

feats_pheno = ['Age', 'Sex', 'Tissue']
pheno = pheno[feats_pheno]

df_clocks = pd.merge(pheno, betas, left_index=True, right_index=True)
df_clocks = pya.pp.epicv2_probe_aggregation(df_clocks, verbose=True)
```

```python
clocks = ["grimage"]

df_clocks['Female'] = (df_clocks['Sex'] == 'F').astype(int)
adata_sex = pya.pp.df_to_adata(df_clocks, metadata_cols=['Sex', 'Tissue'], imputer_strategy='knn', verbose=True)
pya.pred.predict_age(adata=adata_sex, dir=path, clock_names=clocks, verbose=True)
results_sex = pd.merge(pheno.loc[:, feats_pheno], adata_sex.obs[clocks], left_index=True, right_index=True)
```

```python
df_clocks['female'] = (df_clocks['Sex'] == 'F').astype(int)
adata_f = pya.pp.df_to_adata(df_clocks, metadata_cols=['Sex', 'Tissue'], imputer_strategy='knn', verbose=True)
pya.pred.predict_age(adata=adata_f, dir=path, clock_names=clocks, verbose=True)
results_f = pd.merge(pheno.loc[:, feats_pheno], adata_f.obs[clocks], left_index=True, right_index=True)

df_clocks['female'] = (df_clocks['Sex'] == 'F').astype(int)
df_clocks = df_clocks.rename(columns={"Age": "age"})
adata_age = pya.pp.df_to_adata(df_clocks, metadata_cols=['Sex', 'Tissue'], imputer_strategy='knn', verbose=True)
pya.pred.predict_age(adata=adata_age, dir=path, clock_names=clocks, verbose=True)
results_age = pd.merge(pheno.loc[:, feats_pheno], adata_age.obs[clocks], left_index=True, right_index=True)
```
