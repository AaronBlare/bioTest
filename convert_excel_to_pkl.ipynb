```python
import os
import pandas as pd

dir = "E:/YandexDisk/bbd/attacks_on_clocks/result"

for path, folders, files in os.walk(dir):
    for folder_name in folders:
        curr_dir = f"{path}/{folder_name}/Evasion/BasicIterative/"
        for curr_path, curr_folders, curr_files in os.walk(curr_dir):
            for curr_file in curr_files:
                if curr_file.endswith("xlsx"):
                    df = pd.read_excel(f"{curr_path}/{curr_file}", index_col=0)
                    df.to_pickle(f"{curr_path}/{curr_file[:-4]}pkl")
                if curr_file.endswith("csv"):
                    df = pd.read_csv(f"{curr_path}/{curr_file}", index_col=0)
                    df.to_pickle(f"{curr_path}/{curr_file[:-3]}pkl")
```
