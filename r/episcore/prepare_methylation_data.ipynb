```python
import pandas as pd
import pyaging as pya
```

```python
load_data_path = 'E:/YandexDisk/bbd/fmba/dnam/processed/'
data = pd.read_pickle(f'{load_data_path}betas.pkl')
```

```python
data = pya.pp.epicv2_probe_aggregation(data)
```

```python
load_info_path = 'D:/bioTest/r/episcore/'
cpgs_1 = pd.read_csv(f'{load_info_path}cage_coefficients_linear.tsv', sep='\t')
cpgs_2 = pd.read_csv(f'{load_info_path}cage_coefficients_log.tsv', sep='\t')
cpgs_3 = pd.read_csv(f'{load_info_path}Predictors_Shiny_by_Groups.csv')
cpgs_1 = set(cpgs_1['CpG_Site'].to_list())
cpgs_2 = set(cpgs_2['CpG_Site'].to_list())
cpgs_3 = set(cpgs_3['CpG_Site'].to_list())
cpgs = list(cpgs_1.union(cpgs_2, cpgs_3))
cpgs = set([cpg.split('_')[0] if len(cpg.split('_')) > 0 else cpg for cpg in cpgs])
intersected_cpgs = list(set(cpgs).intersection(data.columns))
len(intersected_cpgs)
```

```python
age_sex_info = pd.read_excel(f'{load_data_path}pheno.xlsx', index_col=0)
age_sex_selected = age_sex_info.loc[:, ['Age', 'Sex']]
age_sex_selected['Sex'] = age_sex_selected['Sex'].str.replace('F','Female')
age_sex_selected['Sex'] = age_sex_selected['Sex'].str.replace('M','Male')
age_sex_info.index.name = 'ID'
age_sex_selected.to_csv(f'{load_info_path}age_sex_Les.csv')
```

```python
data_selected = data[intersected_cpgs]
data_t = data_selected.transpose()
data_t.index.name = 'cpg'
data_t.to_csv(f'{load_info_path}data_Les.csv')
```
