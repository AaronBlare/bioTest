```python
import pandas as pd
```

```python
path_load = 'E:/YandexDisk/bbd/fmba/'
df_nir = pd.read_excel(f"{path_load}Таблица_НИР (1).xlsx", index_col=1)
df_pril = pd.read_excel(f"{path_load}ПриложениеГ (КОНФИДЕНЦИАЛЬНО).xlsx", index_col=1)
df_nir = df_nir.drop(columns=['дата начала характеристики работника', 'дата конца характеристики работника'])
df_pril = df_pril.rename(index={7611:49084}) 
df_nir.sort_index(inplace=True)
df_pril.sort_index(inplace=True)
```

```python
comparison_local = df_nir.loc[49084, :].compare(df_pril.loc[49084, :])
```

```python
path_save = 'D:/bioTest/'
with pd.ExcelWriter(f"{path_save}comparison_result.xlsx") as writer:
    all_columns = df_pril.columns
    for curr_column in all_columns:
        curr_comparison = df_nir.loc[:, curr_column].compare(df_pril.loc[:, curr_column], align_axis=1, result_names = ('nir', 'pril'))
        if not curr_comparison.empty:
            curr_comparison.to_excel(writer, sheet_name=curr_column[-29::])  
```
